```python
# Positional arguments -> arguments are passed in the same order as parameters 
def student(name, age):
	print(name, age)
	
student("Ram", 18)
```

```python
# Keyword arguments → arguments are passed using parameter name. Order doesn’t matter
def student(name, age):
	print(name, age)
	
student(age = 20, name = "Ram") # We have different order here
```

```python
# default arguments → parameter have default value
def greet(name="Guest"):
    print("Hello", name)

greet("Ram") 
greet() 
```

```python
# Variable-length arguments → used when number of arguments is not known. 
# *args → variable number of positional arguments 
# **kwargs → variable number of keyword arguments

# hamle positional arguments jati ota pathayem tyo sab chai args ma janxa 
# hamle keyword arguments jati ota pathayem tyo sab chai kwargs ma janxa 

# args saves/stores/collects everything as tuple 
# kwargs saves/stores/collects everything as dictionary
```

```python
def student(*args, **kwargs):
    print("subjects:", args)
    print("details:", kwargs)

student("math", "ai", name="Sita", age= 21)
```

```python
# Local variables are created inside a function and exist only during it's execution.
# Global variables are created outside and exists inside of the funtions too.
def greet():
    msg = "Hello!"
    print("Inside function:", msg)

greet()
print("Outside function:", msg)
```

```python
# Lists 
# Ordered (can get the value from the index)
# Mutable (can change value after creation)
# Duplicates (can have same value multiple times)
# Heterogenous (can have different data types in a list)
lists = [1, "data", 3.14, True, 1]
lists
```

```python
# List
lst = [1, 2, 3]
print(lst)              # Output: [1, 2, 3]

lst.append(4)
print(lst)              # Output: [1, 2, 3, 4]

lst[0] = 10
print(lst)              # Output: [10, 2, 3, 4]

lst.remove(2)
print(lst)              # Output: [10, 3, 4]

x = lst[1]
print(x)                # Output: 3

lst.extend([5, 6])
print(lst)              # Output: [10, 3, 4, 5, 6]

sub = lst[1:3]
print(sub)              # Output: [3, 4]
```

```python
# Set
st = {1, 2, 3}          # create
print(st)

st.add(4)               # add
print(st)

st.remove(2)            # remove
print(st)

x = 3 in st             # membership test
print(x)

st2 = st | {5, 6}       # union
print(st2)

inter = st & {3, 4}     # intersection
print(inter)
```

```python
# Tuple
tpl = (1, 2, 3)         # create
print(tpl)

x = tpl[0]              # access
print(x)

tpl2 = tpl + (4,)       # concatenate
print(tpl2)

a, b, c = tpl           # unpack
print(a, b, c)

cnt = tpl.count(2)      # count value
print(cnt)

idx = tpl.index(3)      # find index
print(idx)
```

```python
# Dictionary 
dct = {"a": 1, "b": 2}  # create
print(dct)

dct["c"] = 3            # add / update
print(dct)

x = dct["a"]            # access by key
print(x)

del dct["b"]            # remove
print(dct)

v = dct.get("x", 99)     # safe access (if the key "x" doesn't exists, it will return 99 i.e. second parameter)
print(v)

item = dct.pop("a")     # remove & return
print(item)
print(dct)
```

```python
# Collections -> container objects that hold and organize data

# Built in collections are lists, tuples, dictionary, etc.
# Python provides a collection module that contains specialized container data types.
# 1. Counter 
# 2. defaultdict
# 3. namedtuple
# 4. deque
```

```python
# Counter counts the frequency of elements, it gives the result in the form of dictionary
from collections import Counter

data = ['a', 'b', 'a', 'c', 'a']
count = Counter(data)
print(count)
```

```python
# defaultdict provides a default value if a key does not exist.
from collections import defaultdict

d = defaultdict(int)
d["a"] += 1
print(d)
```

```python
# namedtuple creates tuple objects with named fields
from collections import namedtuple

Student = namedtuple("Student", ["name", "age"])
s = Student("Ram", 20)

print(s.name)
print(s.age)
```

```python
# deque allows fast insertion and deletion from both ends. It is simply a double ended queue
from collections import deque

dq = deque([1,2,3])
dq.appendleft(0)
dq
```

```python
# Iterators : An object that enables sequential access to it's elements, traversing through all values. 

# To make object iterable, two methods are used: 
# 1. __iter__() -> returns the iterator object
# 2. __next__() -> returns the next element. To indicate the iteration has completed, we use StopIteration

list = [4, 7, 0]
iterator = iter(list)
print(next(iterator))
print(next(iterator))
print(iterator.__next__()) # Other way or writing, but works the same as before
```

```python
# Custom iterators 
class Count:
    def __init__(self):
        self.num = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.num <= 3:
            value = self.num
            self.num += 1
            return value
        else:
            raise StopIteration


c = Count()

for i in c:
    print(i)
```

```python
# Generator : It is simple way to create iterators. 
# They use a simple function and yield keyword instead of the return.

# They are used because:
# Memory efficient : they don't store all values in memory and generate them when needed
# performance : useful for large dataset
# readable code : yield makes complex iteration logic easier to express.

def count():
    i = 1 
    while i<=10:
        yield i
        i += 1

c = count()

for i in c:
    print(i)
```

```python
# Decorators : It is a function that wraps another function adding functionality to it. 
# Or, it is a function that modifies or extends behavious of function without changing the code.
# they are used to modify the behaviour of functions, class, methods.
# they use @ syntax

def decorator_func(func):
    def wrapper():
        print("Before function execution")
        func() 
        print("After function execution")
    return wrapper

@decorator_func
def greet():
    print("Hello")

greet()

# func() calls the original function so hello is in second in output
```

```python
# OOP 
# 1. Class: blueprint for creating objects
# 2. objects : instance of a class
# 3. attributes : variables inside an object
# 4. Methods : functions inside an object

class Dog:  # Creating a class
    def __init__(self, name): # Constructor method to create attribute
        self.name = name # Attribute (data belonging to the object)

    def bark(self): # Defining a method
        print(self.name, "is barking")

d1 = Dog("Tommy")  # Creating an object and giving it a name attribute
d1.bark()  # calling the method
```

```python
# Exception is runtime error that disrupts normal program flow. 
# It prevents program crashes and handles errors and also improves user experience.

# try-except method:
# try -> we will wrap the code which might cause error
# except -> it handles the exception

a = int(input("Enter any number: "))
try:
    print(10 / a)
except:
    print("We can't divide by zero")
    
print("The division is done")
```

```python
# There is finally and it runs regardless of exception (even if there is or isn't any error)
a = int(input("Enter any number: "))

try:
    print(10 / a)
except Exception as err:
    print(f"There is a error as: {err}")
else: 
    print("There is no error")
finally: 
    print("This will always execute")
    
print("The division is done")
```
